```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
predictions = pd.read_csv(
    "../data/processed/model_predictions.csv"
)

metrics = pd.read_csv(
    "../data/processed/model_metrics.csv"
)

predictions["Date"] = pd.to_datetime(
    predictions["Date"]
)

predictions.head()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    predictions["Date"],
    predictions["Actual_ESS_target"],
    label="Actual ESS"
)

plt.plot(
    predictions["Date"],
    predictions["Conventional"],
    label="Conventional Model"
)

plt.plot(
    predictions["Date"],
    predictions["Conventional_Behavioral"],
    label="Conventional + Behavioral"
)

plt.title("Mirai: Actual vs Predicted Economic Stress")
plt.xlabel("Date")
plt.ylabel("Economic Stress Score")
plt.legend()
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    metrics["Model"],
    metrics["RMSE"]
)

plt.title("Model RMSE Comparison")
plt.xlabel("Model")
plt.ylabel("RMSE")

plt.xticks(rotation=15)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    metrics["Model"],
    metrics["MAE"]
)

plt.title("Model MAE Comparison")
plt.xlabel("Model")
plt.ylabel("MAE")

plt.xticks(rotation=15)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    metrics["Model"],
    metrics["MAE"]
)

plt.title("Model MAE Comparison")
plt.xlabel("Model")
plt.ylabel("MAE")

plt.xticks(rotation=15)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    predictions["Date"],
    predictions["Actual_ESS_target"]
)

plt.title("Economic Stress Score Trend")
plt.xlabel("Date")
plt.ylabel("Economic Stress Score")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    predictions["Date"],
    predictions["Actual_ESS_target"],
    label="Actual ESS"
)

plt.plot(
    predictions["Date"],
    predictions["Conventional"],
    label="Conventional Signals"
)

plt.plot(
    predictions["Date"],
    predictions["Conventional_Behavioral"],
    label="Behavioral + Conventional Signals"
)

plt.title(
    "Impact of Behavioral Signals on Economic Stress Forecasting"
)

plt.xlabel("Date")
plt.ylabel("Economic Stress Score")

plt.legend()

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
best_rmse = metrics.loc[
    metrics["RMSE"].idxmin()
]

print("MIRAI VISUALIZATION SUMMARY")
print("-" * 40)

print(
    f"Best model: {best_rmse['Model']}"
)

print(
    f"Best RMSE: {best_rmse['RMSE']:.4f}"
)

print(
    f"Best MAE: {best_rmse['MAE']:.4f}"
)

print(
    f"R² Score: {best_rmse['R2']:.4f}"
)
```
